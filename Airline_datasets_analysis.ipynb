## Flight Analysis

```python
# Number of daily flights (like time series)
```

```python
daily_flights = df['Departure Date'].value_counts().sort_index()

plt.figure(figsize=(14, 5))
sns.lineplot(x=daily_flights.index, y=daily_flights.values)
plt.title("Daily Flight Number (Departure Date)")
plt.xlabel("Date")
plt.ylabel("Number of Flights")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
'''Interpretation: The chart shows fluctuations in daily flight counts, typically ranging between 240 and 310. 
The average appears to be around 270–280 flights per day, with no clear upward or downward trend. 
Regular oscillations may suggest weekly or seasonal cycles, but the dense date range limits precise conclusions.

Business Insight: Understanding daily fluctuations supports better capacity planning and resource allocation. 
Recognizing patterns can also enhance dynamic pricing and operational efficiency, especially during high or low-demand days.'''
```

```python
# Number of flights per month (using the Month column we created earlier)
```

```python
df['Departure Date'] = pd.to_datetime(df['Departure Date'], errors='coerce')

df['Departure Date'] = pd.to_datetime(df['Departure Date'])
df['Month'] = df['Departure Date'].dt.month
```

```python
monthly_flights = df['Month'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
sns.barplot(x=monthly_flights.index, y=monthly_flights.values, palette='coolwarm')
plt.title("Number of Flights by Month")
plt.xlabel("Month")
plt.ylabel("Number of Flights")
plt.show()
```

```python
'''Interpretation: The highest flight volumes occur in March, May, July, August, and October — each peaking around or above 5,000 flights. In contrast, February, April, June, September, November, and December show relatively lower activity.

Seasonality is clearly visible, with peaks in spring, summer, and early autumn. This pattern may reflect holidays, school breaks, or climate preferences.

Business Insight: Airlines can adjust operations seasonally: adding flights and increasing prices in high-demand months while running promotions or optimizing routes during quieter periods. 
This enables efficient resource allocation and revenue maximization.'''
```

```python
# What percentage is on time, delayed, cancelled?
```

```python
status_counts = df['Flight Status'].value_counts()
status_percent = df['Flight Status'].value_counts(normalize=True) * 100

plt.figure(figsize=(6, 4))
sns.barplot(x=status_counts.index, y=status_counts.values, palette='Set3')
plt.title("Flight Status Distribution")
plt.xlabel("Flight Status")
plt.ylabel("Number of Fligts")
plt.show()

print("Flight Status Percentages (%):")
print(status_percent.round(2))
```

```python
'''Interpretation: The chart reveals an almost equal distribution of flight statuses: On Time, Delayed, and Cancelled — each accounting for roughly 32,000 flights. 
This implies that approximately one-third of all flights are either delayed or cancelled, 
which is alarmingly high compared to industry standards (where on-time performance is typically 70–90%).

Business Insight: These rates suggest major operational inefficiencies that could severely impact customer satisfaction,
brand reputation, and financial performance. 
The airline should investigate root causes and prioritize operational optimization through better planning, predictive modeling, and contingency strategies.'''
```

```python
# Which Pilots Have the Most Flight Delays?
```

```python
delayed_flights = df[df['Flight Status'] == 'Delayed']

top_delay_pilots = delayed_flights['Pilot Name'].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=top_delay_pilots.values, y=top_delay_pilots.index, palette='Reds_r')
plt.title("Pilots with the Most Delays (Top 10)")
plt.xlabel("Number of Delayed Flights")
plt.ylabel("Pilot Name")
plt.show()
```

```python
'''Interpretation: The chart shows that no pilot has caused more than two delayed flights — Demetris Atherley and Ethan Desbrow lead with two each,
while the rest have only one. Compared to the previously noted 32,000 delayed flights overall, 
this suggests that pilot-related delays are minimal and not the root cause of systemic delays.

Business Insight: Pilot performance appears consistent, with no major outliers. 
The airline should shift its attention to system-level issues (weather, aircraft maintenance, scheduling) 
when addressing overall delay reduction.'''
```

```python
pilot_flight_counts = df.groupby('Pilot Name')['Flight Status'].value_counts(normalize=True).unstack().fillna(0)
pilot_delay_rate = pilot_flight_counts['Delayed'].sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=pilot_delay_rate.values, y=pilot_delay_rate.index, palette='Reds')
plt.title("Pilots with the Highest Delay Rates (Top 10)")
plt.xlabel("Delay Rate")
plt.ylabel("Pilot Name")
plt.show()
```

```python
'''Interpretation: All pilots shown have a delay rate of 1.0 or very close to it, meaning all their recorded flights were delayed. 
This is highly abnormal in airline operations. 
Possible explanations include small sample sizes (e.g., a pilot with one delayed flight = 100% delay), 
data recording or processing errors, unusual delay definitions, or pilots assigned to high-risk or special operations.

Business Insight: This insight is valuable for both data validation and performance management. 
If valid, these outlier pilots should be reviewed in the context of their routes and assignments.
If not, improvements in data quality control are necessary to maintain accurate operational decision-making.'''
```

```python
# Which Nations Fly More in Which Months?
```

```python
df['Departure Date'] = pd.to_datetime(df['Departure Date'], errors='coerce')
df['Month'] = df['Departure Date'].dt.month
```

```python
monthly_nationality = df.groupby(['Month', 'Nationality']).size().reset_index(name='Passenger Count')
```

```python
top_nationalities = df['Nationality'].value_counts().head(5).index
filtered_data = monthly_nationality[monthly_nationality['Nationality'].isin(top_nationalities)]

plt.figure(figsize=(12, 6))
sns.lineplot(data=filtered_data, x='Month', y='Passenger Count', hue='Nationality', marker='o')
plt.title("Monthly Passenger Counts by Nationality (Top 5)")
plt.xlabel("Month")
plt.ylabel("Passenger Count")
plt.xticks(range(1,13))
plt.grid(True)
plt.show()
```

```python
'''Interpretation: China overwhelmingly leads in passenger volume, marking it as a key market segment. 
Indonesia, the Philippines, and Russia show relatively stable passenger numbers throughout the year. 
Brazil has the lowest passenger volume among the top five, indicating less traffic from this nationality. 
Notable seasonal fluctuations in Chinese passenger numbers suggest travel habits are influenced by holidays or special events.

Business Insight: Seasonal promotions and capacity planning are crucial for the Chinese market. 
Stable markets require consistent service and loyalty initiatives. 
Opportunities to expand in lower-traffic countries like Brazil should be explored, with flexible scheduling to manage seasonality effectively.'''
```
